```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import polars as pl

from sometria.representation import Representation, _load_human_definition
from sometria.normalization import save_feature_normalization
from sometria.catalog import (
    MotionViewSpec,
    build_motion_view,
    create_pretrain_split,
    load_catalog,
    load_splits,
)
```

```python
OUT = Path("../data/processed")
HUMAN = _load_human_definition("../config/human.yaml")
REPRESENTATION = Representation.from_human(HUMAN)
```

```python
# Create pretrain split with only AMASS

create_pretrain_split(
    output_root=OUT,
    source_datasets=("AMASS",),
    split_set="pretrain_v1",
)
```

```python
# Create view and normalization stats for pretrain v1 clean

spec = MotionViewSpec(
    source_datasets=("AMASS",),
    split_set="pretrain_v1", 
    min_frames=240,
    split="train",
)

samples = build_motion_view(OUT, spec)
print("pretrain_v1_stats_train_clean", f"{len(samples):,} samples")
save_feature_normalization(
    output_root=OUT,
    samples=samples,
    name="pretrain_v1_stats_train_clean",
    representation=REPRESENTATION,
    spec=spec,
)
```

```python
# Create pretrain split with AMASS and MotionX

create_pretrain_split(
    output_root=OUT,
    source_datasets=("AMASS", "MotionX"),
    split_set="pretrain_v2",
)
```

```python
# Create view and normalization stats for pretrain v2 clean

spec = MotionViewSpec(
    source_datasets=("AMASS", "MotionX"),
    split_set="pretrain_v2", 
    min_frames=240,
    split="train",
)

samples = build_motion_view(OUT, spec)
print("pretrain_v2_stats_train_clean", f"{len(samples):,} samples")
save_feature_normalization(
    output_root=OUT,
    samples=samples,
    name="pretrain_v2_stats_train_clean",
    representation=REPRESENTATION,
    spec=spec
)
```

```python
# Create view and normalization stats for pretrain v2 complete

spec = MotionViewSpec(
    source_datasets=("AMASS", "MotionX"),
    split_set="pretrain_v2", 
    exclude_broken=False,
    min_frames=240,
    split="train",
)

samples = build_motion_view(OUT, spec)
print("pretrain_v2_stats_train_complete", f"{len(samples):,} samples")
save_feature_normalization(
    output_root=OUT,
    samples=samples,
    name="pretrain_v2_stats_train_complete",
    representation=REPRESENTATION,
    spec=spec
)
```
